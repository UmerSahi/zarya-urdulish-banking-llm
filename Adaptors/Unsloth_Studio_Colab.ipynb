```python
import os
import shutil

# Define the target folder name
folder_name = "/content/saved_model_files"
os.makedirs(folder_name, exist_ok=True)

# List of files to move/copy
files_to_save = [
    "tokenizer_config.json",
    "chat_template.jinja",
    "export_metadata.json",
    "adapter_config.json",
    "adapter_model.safetensors",
    "tokenizer.json"
]

# Copy each file to the new folder if it exists in /content/
for file in files_to_save:
    source_path = os.path.join("/content", file)
    if os.path.exists(source_path):
        dest_path = os.path.join(folder_name, file)
        shutil.copy(source_path, dest_path)
        print(f"Copied: {file} -> {folder_name}/")
    else:
        print(f"File not found: {file}")

print("\nDone! All available selected files have been saved to:", folder_name)
```

```python
import shutil
from google.colab import files

# Compress the saved_model_files directory into a zip archive
shutil.make_archive('/content/saved_model_files', 'zip', '/content/saved_model_files')

# Trigger the browser download for the zip file
files.download('/content/saved_model_files.zip')
```

```python
import os
import shutil
from google.colab import files

# Let's check if 'llama-adapters' or any similar directory exists
target_dir = '/content/llama-adapters'

# If it doesn't exist under that exact name, let's look around for anything with 'adapter' in it
if not os.path.exists(target_dir):
    possible_dirs = [d for d in os.listdir('/content') if os.path.isdir(os.path.join('/content', d)) and 'adapter' in d.lower()]
    if possible_dirs:
        target_dir = os.path.join('/content', possible_dirs[0])
        print(f"Found matching folder: {target_dir}")

if os.path.exists(target_dir):
    zip_path = f"{target_dir}_download"
    shutil.make_archive(zip_path, 'zip', target_dir)
    print(f"Zipped {target_dir} to {zip_path}.zip. Starting download...")
    files.download(f"{zip_path}.zip")
else:
    print("Could not find a 'llama-adapters' folder. Please make sure the folder exists or check the name.")
```
